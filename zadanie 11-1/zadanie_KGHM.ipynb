## Zadanie: ceny akcji

Stwórz dwa osobne wykresy liniowe (jeden pod drugim) – jeden niech przedstawia ceny zamknięcia KGH, drugi niech przedstawia ceny miedzi (na podstawie kolumn "Zamknięcie"). Jako trzeci element wykresu dodaj tabelę zestawiającą ceny zamknięcia KGHM oraz miedzi w kolejnych dniach (ponownie na podstawie kolumny "Zamknięcie").

Import niezbednych bibliotek

```python
import pandas as pd
import plotly.graph_objects as go
from plotly.subplots import make_subplots
```

Wczytanie danych z plików CSV oraz zamiana typu na date

```python
df_KGH = pd.read_csv('kgh_d.csv')
df_KGH['Data'] = pd.to_datetime(df_KGH['Data'])
df_miedz = pd.read_csv('ca_c_f_d.csv')
df_miedz['Data'] = pd.to_datetime(df_miedz['Data'])
```

Utworzenie df zawierajacego niezbedne dane do utworzenia wykresów i tabeli, usuniecie niepotrzebnych kolumn oraz zmaiana nazw na odpowiadajace zadaniu.

```python
df_table = pd.merge(df_KGH, df_miedz, on='Data', how='inner')
df_table = df_table[['Data', 'Zamkniecie_x', 'Zamkniecie_y']]
df_table = df_table.rename(columns={'Zamkniecie_x': 'KGH', 'Zamkniecie_y': 'Miedź'})
```

Tworzenie 2 wykresów i tabeli na jednym "płótnie"

```python

x = df_table['Data']
y = df_table['KGH']
y2 = df_table['Miedź']


fig = make_subplots(
    rows=3, 
    cols=1,
    shared_xaxes= True,
    vertical_spacing=0.1,
    subplot_titles=('KGH - Cena zamknięcie', 'Miedz - Cena zamknięcie', 'Tabela wartości'),
    specs=[[{"type": "xy"}],
            [{"type": "xy"}],  
            [{"type": "table"}]]
)

fig.add_trace(
    go.Scatter(x=x, y=y, name='KGH', line=dict(color='blue')),
    row=1, col=1
)

fig.add_trace(
    go.Scatter(x=x, y=y2, name='Miedz', mode='lines'),
    row=2, col=1
)
fig.add_trace(
    go.Table(
        header=dict(
            values=list(df_table.columns),
            align='left'
        ),
        cells=dict(
            values=[df_table[col] for col in df_table.columns],
            align='left'
        )
    ),
    row=3, col=1
)

fig.update_layout(
    title='Wartość akcji KGH oraz wartość miedzi',
    title_x=0.5,
    height=1000,
    showlegend=True,
    hovermode="x unified"
)

```
